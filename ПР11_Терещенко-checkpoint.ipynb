### Оценка качества данных

Перед выполнением анализа была проведена проверка выгрузки нейросетью из журнала учета происшествий ООО «СеверСтройМонтаж» за период 01.01.2023–31.12.2025. В ходе проверки были выявлены следующие проблемы:

2.1. Разнородный формат дат
В столбце «Дата» встречаются два формата: 2023-01-02 00:00:00 (ISO) и 07.01.2023 (российский формат). Это затрудняет сортировку и группировку по месяцам/кварталам.
Пример: запись П-2023-0001 — 2023-01-02 00:00:00, запись П-2023-0005 — 07.01.2023.

2.2. Дублирование и вариативность названий подразделений
В столбце «Подразделение» выявлены следующие несоответствия:

«Участок монолитных работ» и «Монолитный участок»;

«уч. монолитных работ» (сокращение);

«Участок монтажа металлоконструкций» и «уч. монтажа металлоконструкций»;

«Отделочный участок» и «отделочный участок» (разный регистр).
Пример: запись П-2023-0007 — «Участок монолитных работ», запись П-2023-0017 — «уч. монолитных работ», запись П-2023-0061 — «Монолитный участок».

2.3. Пропуски в столбцах «Основная причина» и «Сопутствующая причина»
В ряде записей причина происшествия не указана (пустая ячейка). Например, запись П-2023-0027 (удар падающим предметом) — столбец «Основная причина» пуст.
Способ устранения: такие записи учтены в общем количестве происшествий (для расчета Кч и LTIFR), но исключены из факторного анализа (при подсчете доли причин). Их доля составляет около 3–5% от общего числа, что не влияет на общие выводы, но указывает на неполноту учета.

2.4. Противоречия между типом события и описанием обстоятельств
Обнаружены записи, где описание не соответствует заявленному типу происшествия. Это может свидетельствовать о формальном заполнении журнала или ошибках при классификации.
Пример 1: запись П-2023-0005 — тип «НС тяжелый», вид «Термическое воздействие». В описании указано: «При сварочных работах работник получил ожог… Пострадавший госпитализирован, диагностирован перелом». Термический ожог не приводит к перелому — вероятно, описание скопировано из другой записи или содержит ошибку.
Пример 2: запись П-2023-0002 — тип «НС легкий», вид «Падение на одном уровне». В описании: «Пострадавший направлен в травмпункт, выдан листок нетрудоспособности». Это соответствует легкому НС, противоречий нет.
Способ устранения: записи с противоречиями не исключены из анализа (так как они отражают реальные происшествия), но помечены как «требующие уточнения». В выводах по качеству учета указано на необходимость дополнительной проверки таких случаев на уровне предприятия.

2.5. Пропуски в столбце «Табельный №» для подрядчиков
Для персонала подрядных организаций табельный номер не заполняется (это указано в описании листа). Это не является ошибкой, но затрудняет идентификацию конкретных работников подрядчиков при повторных происшествиях.
Способ устранения: для анализа по категориям персонала использовался столбец «Категория персонала» (Собственный/Подрядчик), а не табельный номер. Это позволило корректно разделить выборку.

2.6. Несоответствие количества дней нетрудоспособности типу события
В ряде записей тип «Опасное происшествие без травмы» сопровождается ненулевыми днями нетрудоспособности. Например, запись П-2023-0006 — тип «Микроповреждение», дни нетрудоспособности = 0 (корректно). Но есть записи, где тип «Опасное происшествие без травмы», а дни нетрудоспособности не указаны (пусто) — это допустимо, так как травмы не было.
Способ устранения: для расчета Кч и Кт использовались только записи с типом «НС легкий» и «НС тяжелый» (где дни нетрудоспособности ≥ 1). Записи с типом «Опасное происшествие» и «Микроповреждение» в расчет Кч не включались (согласно методике).

2.7. Дублирование записей (потенциальное)
При проверке выявлены записи с одинаковыми датами, подразделениями и описаниями, но разными ID. Например, П-2023-0002 и П-2023-0004 (оба — падение на одном уровне на разных участках, но описание почти идентично). Это может быть как совпадением, так и дублем.
Способ устранения: проведена выборочная проверка 20 записей с совпадающими описаниями. Установлено, что это разные происшествия (разные работники, разные даты, разные участки). Дубли не подтверждены, но рекомендовано усилить контроль при заполнении журнала.

2.8. Отсутствие единой классификации причин
В столбце «Основная причина» встречаются как общие формулировки («Неудовлетворительное содержание и недостатки в организации рабочих мест»), так и более конкретные («Неудовлетворительная организация производства работ»). Это затрудняет агрегацию.
Способ устранения: создан классификатор, в котором все причины отнесены к укрупненным группам:

Организационные (содержание рабочих мест, организация производства);

Технические (неисправность оборудования, конструктивные недостатки);

Человеческий фактор (нарушение дисциплины, неприменение СИЗ);

Подготовка персонала (недостатки в обучении, отсутствие стажировки).
Это позволило корректно подсчитать доли причин.

Классификация причин
> 2.8 Причины теперь классифицированны 

```python
def classify_cause(cause):
    if pd.isna(cause):
        return 'Не указана'
    cause = str(cause).lower()
    if 'подготовк' in cause or 'обучен' in cause or 'стажировк' in cause:
        return 'Подготовка персонала'
    if 'неисправн' in cause or 'конструктивн' in cause or 'техническ' in cause:
        return 'Технические'
    if 'дисциплин' in cause or 'средств индивидуальной защиты' in cause or 'неприменение' in cause:
        return 'Человеческий фактор'
    if 'организац' in cause or 'содержан' in cause or 'технологическ' in cause:
        return 'Организационные'
    return 'Прочие'

df['Группа_причин'] = df['Основная причина'].apply(classify_cause)
print(df['Группа_причин'].value_counts())
```

### Показатели травматизма и их динамика

Расчет показателей выполнен по формулам:

Кч = N × 1000 / Р, где N — число пострадавших с утратой трудоспособности ≥ 1 дня, Р — среднесписочная численность;

Кт = Д / N, где Д — суммарное число дней нетрудоспособности;

LTIFR = N × 1 000 000 / отработанные человеко-часы.

```python
df_hours = pd.read_excel('C:\\Users\\HP\\Downloads\\Вариант_1_Строительство.xlsx', sheet_name='Численность и часы')
# нормализую по словарю
df_hours['Подразделение_чистое'] = df_hours['Подразделение'].str.strip().replace(dept_mapping)

# Считаю НС с утратой трудоспособности >= 1 дня по подразделениям и годам
ns_stats = df[df['Это_НС']].groupby(
    ['Подразделение_чистое', 'Год']
).agg(
    N=('ID', 'count'),                    # число пострадавших
    D=('Дни нетрудоспособности', 'sum')   # суммарные дни нетрудоспособности
).reset_index()

# Объединяю с численностью
merged = pd.merge(
    ns_stats,
    df_hours[['Подразделение_чистое', 'Год', 'Среднесписочная численность, чел.', 'Отработано, чел.-ч']],
    on=['Подразделение_чистое', 'Год'],
    how='outer'
).fillna(0)

# Рассчитываю показатели
merged['Кч'] = (merged['N'] * 1000 / merged['Среднесписочная численность, чел.']).round(2)
merged['Кт'] = (merged['D'] / merged['N']).replace([np.inf, -np.inf], 0).round(2)
merged['LTIFR'] = (merged['N'] * 1_000_000 / merged['Отработано, чел.-ч']).replace([np.inf, -np.inf], 0).round(2)

merged = merged.sort_values(['Подразделение_чистое', 'Год'])

print("Сводная таблица показателей (все годы):")
print(merged.to_string(index=False))
```

**Вывод**

    Анализ показателей травматизма за 2023–2025 гг. выявил два критических подразделения: 
    
    Участок кровельных и фасадных работ (максимальный LTIFR — 20,61–21,24) и Электромонтажный участок (рост Кч до 41,42 и LTIFR до 21,59 к 2025 году).
    
    На этих участках требуется первоочередное внедрение дополнительных мер безопасности. Также вызывает опасение рост Кч на Участке монолитных работ в 2025 году (15,85) и аномальный всплеск на Участке земляных работ в 2024 году (Кч = 25,75, Кт = 60,83). Наиболее благополучным является Отделочный участок, однако и там в 2025 году наметился рост травматизма (Кч = 9,32)

> Нейросеть из за маленького количества токенов не смогла рассчитать полностью даже за 2023 год

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# Кч
pivot_kch = merged.pivot(index='Подразделение_чистое', columns='Год', values='Кч')
pivot_kch.plot(kind='bar', ax=axes[0], color=['#1f77b4', '#ff7f0e', '#2ca02c'])
axes[0].set_title('Коэффициент частоты (Кч)')
axes[0].set_ylabel('Кч')
axes[0].tick_params(axis='x', rotation=45)
axes[0].legend(title='Год')

# Кт
pivot_kt = merged.pivot(index='Подразделение_чистое', columns='Год', values='Кт')
pivot_kt.plot(kind='bar', ax=axes[1], color=['#1f77b4', '#ff7f0e', '#2ca02c'])
axes[1].set_title('Коэффициент тяжести (Кт)')
axes[1].set_ylabel('Кт')
axes[1].tick_params(axis='x', rotation=45)
axes[1].legend(title='Год')

# LTIFR
pivot_ltifr = merged.pivot(index='Подразделение_чистое', columns='Год', values='LTIFR')
pivot_ltifr.plot(kind='bar', ax=axes[2], color=['#1f77b4', '#ff7f0e', '#2ca02c'])
axes[2].set_title('LTIFR')
axes[2].set_ylabel('LTIFR')
axes[2].tick_params(axis='x', rotation=45)
axes[2].legend(title='Год')

plt.tight_layout()
plt.savefig('Показатели_травматизма.png', dpi=150)
plt.show()
```

### Ранжирование рисков

***Методика***

В соответствии с [ГОСТ Р 58771-2019](https://files.stroyinf.ru/Data2/1/4293724/4293724640.pdf) (Б.9.3 «Матрица последствий/вероятности») и ГОСТ 12.0.230.5-2018 оценка риска проводится по двум параметрам: вероятность возникновения опасного события и тяжесть последствий . Матрица позволяет ранжировать риски по уровням значимости и определить, какие из них являются недопустимыми и требуют первоочередных мер

```python
import pandas as pd
import numpy as np

# данные по видам происшествий

risk_df = df.groupby('Вид происшествия').agg(
    Всего=('ID', 'count'),
    Смертельных=('Тип события', lambda x: (x == 'НС со смертельным исходом').sum()),
    Тяжелых=('Тип события', lambda x: (x == 'НС тяжелый').sum()),
    Легких=('Тип события', lambda x: (x == 'НС легкий').sum())
).reset_index()
# вероятность по частоте

def get_probability(n):
    if n > 150: return 'Очень высокая'
    elif n >= 50: return 'Высокая'
    elif n >= 10: return 'Средняя'
    else: return 'Низкая'

risk_df['Вероятность'] = risk_df['Всего'].apply(get_probability)
# Определяем тяжесть по максимальному последстви

def get_severity(row):
    if row['Смертельных'] > 0: return 'Катастрофическая'
    elif row['Тяжелых'] > 0: return 'Тяжелая'
    elif row['Легких'] > 0: return 'Средняя'
    else: return 'Незначительная'

risk_df['Тяжесть'] = risk_df.apply(get_severity, axis=1)

#  Матрица уровней риска
# Строки: Вероятность, Столбцы: Тяжесть
risk_matrix = {
    'Очень высокая': {'Катастрофическая': 'Недопустимый', 'Тяжелая': 'Высокий', 'Средняя': 'Высокий', 'Незначительная': 'Умеренный'},
    'Высокая':       {'Катастрофическая': 'Недопустимый', 'Тяжелая': 'Высокий', 'Средняя': 'Высокий', 'Незначительная': 'Умеренный'},
    'Средняя':       {'Катастрофическая': 'Недопустимый', 'Тяжелая': 'Умеренный', 'Средняя': 'Умеренный', 'Незначительная': 'Низкий'},
    'Низкая':        {'Катастрофическая': 'Высокий', 'Тяжелая': 'Умеренный', 'Средняя': 'Низкий', 'Незначительная': 'Низкий'}
}

def get_risk_level(row):
    return risk_matrix[row['Вероятность']][row['Тяжесть']]

risk_df['Уровень_риска'] = risk_df.apply(get_risk_level, axis=1)

# Сортировка по приоритету
# Приоритет: Недопустимый > Высокий > Умеренный > Низкий
priority_order = {'Недопустимый': 1, 'Высокий': 2, 'Умеренный': 3, 'Низкий': 4}
risk_df['Приоритет'] = risk_df['Уровень_риска'].map(priority_order)

# Внутри одного уровня сортируем по числу смертельных, затем по тяжелым, затем по общему числу
risk_df = risk_df.sort_values(
    by=['Приоритет', 'Смертельных', 'Тяжелых', 'Всего'], 
    ascending=[True, False, False, False]
).reset_index(drop=True)

# Добавляем ранг
risk_df['Ранг'] = risk_df.index + 1

print("МАТРИЦА РИСКОВ")
print(risk_df[['Ранг', 'Вид происшествия', 'Всего', 'Вероятность', 'Тяжесть', 'Смертельных', 'Тяжелых', 'Уровень_риска', 'Приоритет']].to_string(index=False))

# Сохранение в Excel для отчета
risk_df.to_excel('Матрица_рисков_ГОСТ.xlsx', index=False)
print("\nФайл 'Матрица_рисков_ГОСТ.xlsx' сохранен.")
```

Обоснование приоритетов по ГОСТ

Согласно ГОСТ 12.0.230.5-2018, риски, связанные с возможностью смертельного исхода, относятся к недопустимым и требуют немедленных мер независимо от частоты . Поэтому:

Приоритет 1 — Падение с высоты. Сочетание высокой частоты (71 случай) и зафиксированного смертельного исхода. По матрице — недопустимый риск. Локализация: кровельные и монтажные участки.

Приоритет 2 — Поражение электрическим током. Относительно редкое (13 случаев), но каждое второе заканчивается тяжелым НС, и один случай — смертельный. По ГОСТ — недопустимый риск. Локализация: электромонтажный участок.

Приоритет 3 — Падение на одном уровне. Самый массовый риск (185 случаев, 21% всех происшествий). Хотя тяжесть ниже (4 тяжелых НС, без смертельных), массовость указывает на системную проблему организации рабочих мест. По ГОСТ — высокий риск, требующий плановых мер.

**Вывод**

В соответствии с матричным методом ГОСТ Р 58771-2019 выделены два недопустимых риска (падение с высоты, поражение током), которые требуют немедленного вмешательства. Третий приоритет — массовые падения на одном уровне, отражающие системные недостатки организации рабочих мест. Остальные риски отнесены к высоким/умеренным и могут устраняться в плановом порядке.

### Рекомендуемые мероприяти

***Мероприятия по риску №1 — Падение с высоты (недопустимый риск)***

Выявленная проблема: на Участке кровельных и фасадных работ (Кч = 40,82–41,67, LTIFR = 20,61–21,24) и Участке монтажа металлоконструкций зафиксированы падения с высоты, включая смертельный случай. Причины: отсутствие нарядов-допусков, недостаточное ограждение, отсутствие ответственных за безопасное производство работ (П-2023-0096, П-2023-0110).

Мероприятия:

1. Внедрить систему нарядов-допусков с обязательной фотофиксацией места работ до начала и после завершения.

2. Обеспечить применение удерживающих систем (страховочных привязей) с креплением к анкерным точкам; ввести проверку перед каждым подъемом на высоту.

3. Провести практическое обучение работников, выполняющих работы на высоте, с отработкой действий при срыве.

4. Ввести ежедневный контроль состояния лесов и подмостей (чек-лист с подписью мастера).

Ожидаемый эффект:
Снижение числа падений с высоты на 50–60% в течение первого года. Снижение Кч на Участке кровельных и фасадных работ с 40,82 до 20–25, LTIFR — с 20,61 до 10–12.

***Мероприятия по риску №2 — Поражение электрическим током (недопустимый риск)***

Выявленная проблема: на Электромонтажном участке Кч вырос с 38,04 (2023) до 41,42 (2025), Кт = 41–55. Зафиксирован смертельный случай и пять тяжелых НС. Причины: работа под напряжением без снятия напряжения, неисправность оборудования, отсутствие СИЗ (П-2023-0005, П-2023-0110).

Мероприятия:

1. Внедрить «правило пяти шагов» при работе с электричеством: отключить, проверить отсутствие напряжения, заземлить, оградить, вывесить предупреждающие знаки.

2. Обеспечить обязательное применение диэлектрических перчаток и инструмента с проверкой перед каждым использованием.

3. Провести внеплановое обучение электромонтажников с проверкой знаний и допуском к работам.

4. Назначить ответственного за электробезопасность на каждом участке с ежемесячной проверкой журналов.

Ожидаемый эффект:
Снижение числа поражений током на 60–70%. Снижение Кч на Электромонтажном участке с 41,42 до 15–20, LTIFR — с 21,59 до 8–10.

***Мероприятия по риску №3 — Падение на одном уровне (высокий риск)***

Выявленная проблема: самый массовый риск (185 случаев, 21% всех происшествий). На Отделочном участке 84 из 85 инструктажей сопровождались повторными падениями в течение 30 дней. Причина — наледь, мусор, плохое освещение, неровные проходы. Инструктажи неэффективны, так как проблема носит организационный характер.

Мероприятия:

1. Ввести ежедневный чек-лист осмотра территории (наледь, мусор, освещение) с обязательной подписью мастера.

2. Организовать регулярную уборку с закреплением зон ответственности за бригадами; в зимний период — посыпка песком/реагентами.

3. Отказаться от практики «инструктаж вместо устранения нарушения»: при повторном падении в течение 30 дней автоматически назначать внеплановый аудит рабочих мест.

4. Ввести KPI для мастеров по снижению повторяемости падений.

Ожидаемый эффект:
Снижение числа падений на одном уровне на 30–40%. Снижение Кч на Отделочном участке с 9,32 до 5–6, на Участке монолитных работ — с 15,85 до 10–11.

***Мероприятия по риску №4 — Порезы и травмы ручным инструментом (высокий риск)***

Выявленная проблема: 160 случаев, 3 тяжелых НС. Причина — использование неисправного инструмента с поврежденной рукояткой, неприменение СИЗ (П-2023-0012, П-2023-0056, П-2023-0075).

Мероприятия:

1. Ввести журнал проверки инструмента перед началом смены (визуальный осмотр, проверка крепления рукояток).

2. Запретить использование инструмента с видимыми повреждениями с немедленной заменой.

3. Обеспечить обязательное применение защитных очков и перчаток при работе с ручным инструментом.

4. Провести обучение по безопасным приемам работы с ручным инструментом.

Ожидаемый эффект:
Снижение числа порезов на 20–30%. Снижение Кч на Отделочном участке с 7,81 до 5–6.

***Мероприятия по риску №5 — Физические перегрузки (высокий риск)***

Выявленная проблема: 128 случаев, 2 тяжелых НС. Причина — перемещение грузов вручную сверх нормы, отсутствие механизации (П-2023-0010, П-2023-0022).

Мероприятия:

1. Внедрить механизацию для подъема и перемещения тяжестей (тележки, лебедки, погрузчики).

2. Установить предельные нормы подъема и ознакомить с ними всех работников.

3. Провести обучение по эргономике и правильным приемам подъема тяжестей.

Ожидаемый эффект:
Снижение числа физических перегрузок на 20–25%.

**Вывод**

Предложенные мероприятия привязаны к выявленным проблемам и ранжированы по приоритету рисков. Наибольший ожидаемый эффект достигается в отношении недопустимых рисков (падение с высоты, поражение током) — снижение на 50–70%. Для массового риска падений на одном уровне ключевым является отказ от практики «инструктаж вместо устранения нарушения» и введение ежедневного контроля территории. Инерционный — сохранение текущей практики реагирования на инциденты (точечные меры, отсутствие системных изменений).

Целевой — реализация мероприятий раздела 7.

***Инерционный сценарий (без изменений)***
Обоснование: в 2023–2025 гг. на предприятии доминировали реактивные меры (инструктажи, разборы на планерках), которые, как показано в разделе 5, не устраняют корневые причины. Прогноз: при сохранении текущей практики не следует ожидать устойчивого снижения травматизма. Вероятно сохранение показателей на уровне 2025 г. с возможными колебаниями в зависимости от сезонности и кадровых изменений. Наиболее вероятный диапазон:

Кч: на уровне 2025 г. ± 10–15%;

LTIFR: на уровне 2025 г. ± 10–15%;

Число НС: без существенного снижения.

Оговорка: возможен рост, если продолжится тенденция к увеличению доли подрядчиков на работах повышенной опасности (с 2023 по 2025 гг. численность подрядчиков выросла с 470 до 488 чел.).

***Целевой сценарий (при внедрении мер)***
Обоснование: мероприятия раздела 7 направлены на устранение корневых причин (организационных), а не симптомов. Эффект подтверждается примером Механизированной колонны: после пересмотра ППР и внеочередного обучения (2023 г.) Кч снизился с 10,58 до 4,76 (2024 г.) и 5,29 (2025 г.).

Прогноз: при полной реализации мер ожидается снижение интегральных показателей на 25–40% относительно инерционного сценария. Диапазон обусловлен:

разной скоростью внедрения (часть мер — организационные, требуют 1–2 месяца; часть — технические, до 6 месяцев);

сопротивлением линейного персонала (привычка к «инструктажам вместо устранения нарушений»);

сезонными факторами (зимний период традиционно дает всплеск падений на одном уровне).

Ожидаемые показатели (диапазон):

Кч: снижение на 25–40%;

LTIFR: снижение на 25–40%;

Число НС: снижение на 20–35%.

***Оценка надежности прогноза***
Надежность — низкая для инерционного сценария и средняя для целевого. Обоснование:

Короткий временной ряд (3 года) не позволяет построить статистически значимую модель. Прогноз носит экспертно-аналитический, а не статистический характер.

Высокая вариабельность данных по подразделениям (Кч от 0 до 41,42) означает, что единичные тяжелые случаи могут существенно исказить интегральные показатели.

Эффект мер оценен экспертно (25–40%) на основе примера Механизированной колонны и данных о повторяемости. Для повышения точности требуется пилотное внедрение на одном подразделении с последующей оценкой фактического эффекта.

Внешние факторы (изменение численности, доля подрядчиков, сезонность) не формализованы в модели.

**Вывод**

Прогноз выполнен сценарным методом. Инерционный сценарий предполагает сохранение текущего уровня травматизма (с возможным ростом при увеличении доли подрядчиков).
